```python
import pandas as pd
import numpy as np
```

```python
# Read the original 311 Dataset
data = pd.read_csv("../Data/311_Dataset.csv")
data.shape
```

```python
# Sort the dataset by 'create_date_et' in descending order
data_sorted = data.sort_values(by='create_date_et', ascending=False)
data_sorted.head()
```

```python
# Filter the dataset to include only the last 5 years of data
data_filtered = data[data['create_date_et'].str[:4].astype(int).isin([2025, 2024, 2023, 2022, 2021])]
data_filtered.head()
```

```python
# Shape of the filtered dataset
data_filtered.shape
```

```python
# Convert filtered data to a CSV file
data_filtered.to_csv("../Data/Filtered_311_Dataset.csv", index=False)
```
